```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

file_path = 'Data.xlsx'
df = pd.read_csv(file_path)

cols_to_plot = [
    'A1. Giới tính:',
    'A2. Ngành học:',
    'A3. Sinh viên:',
    'A4. Học lực học kỳ gần nhất:',
    'B1. Tổng thời lượng sử dụng trong ngày:',
    'B2. Khoảng thời gian sử dụng trong ngày:',
    'B3. Bạn thường sử dụng Facebook cho mục đích:',
    'B4. Phương tiện, thiết bị sử dụng:'
]

def get_counts(series):
    if series.dtype == 'object' and series.str.contains(',').any():
        return series.dropna().astype(str).str.split(',').explode().str.strip().value_counts()
    else:
        return series.value_counts()

fig, axes = plt.subplots(4, 2, figsize=(15, 20))
axes = axes.flatten()

for i, col in enumerate(cols_to_plot):
    counts = get_counts(df[col])
    counts = counts.sort_index()
    
    sns.barplot(x=counts.values, y=counts.index, ax=axes[i], orient='h', palette='viridis')
    
    short_title = col.split(':')[0]
    axes[i].set_title(f'Phân bố biến {short_title}')
    axes[i].set_xlabel('Số lượng')
    axes[i].set_ylabel('')

plt.tight_layout()
plt.show()
```
