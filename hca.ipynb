```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("Data.xlsx")

def plot_categorical_by_index(df, col_index, title):
    data = df.iloc[:, col_index].value_counts()
    plt.figure()
    data.plot(kind="bar")
    plt.title(title)
    plt.xlabel("Nhóm")
    plt.ylabel("Số lượng")
    plt.xticks(rotation=45, ha="right")
    plt.tight_layout()
    plt.show()

plot_categorical_by_index(df, 0, "A1. Giới tính")
plot_categorical_by_index(df, 1, "A2. Ngành học")
plot_categorical_by_index(df, 2, "A3. Sinh viên")
plot_categorical_by_index(df, 3, "A4. Học lực học kỳ gần nhất")
plot_categorical_by_index(df, 4, "B1. Tổng thời lượng sử dụng trong ngày")
plot_categorical_by_index(df, 5, "B2. Khoảng thời gian sử dụng trong ngày")
plot_categorical_by_index(df, 6, "B3. Mục đích sử dụng Facebook")
plot_categorical_by_index(df, 7, "B4. Phương tiện, thiết bị sử dụng")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("Data.xlsx")

def group_small(series, min_ratio=0.05):
    ratio = series.value_counts(normalize=True)
    keep = ratio[ratio >= min_ratio].index
    return series.apply(lambda x: x if x in keep else "Khác")


# A1 – Giới tính (pie)
plt.figure(figsize=(6, 6))
a1 = group_small(df.iloc[:, 0])
a1.value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("A1. Giới tính")
plt.ylabel("")
plt.tight_layout()
plt.savefig("A1_gioi_tinh.png", dpi=300, bbox_inches="tight")
plt.show()


# A4 – Học lực (pie)
plt.figure(figsize=(6, 6))
a4 = group_small(df.iloc[:, 3])
a4.value_counts().plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("A4. Học lực học kỳ gần nhất")
plt.ylabel("")
plt.tight_layout()
plt.savefig("A4_hoc_luc.png", dpi=300, bbox_inches="tight")
plt.show()


# A2 – Ngành học (barh + số lượng)
plt.figure(figsize=(8, 6))
a2 = group_small(df.iloc[:, 1]).value_counts().sort_values()
bars = plt.barh(a2.index, a2.values)
plt.title("A2. Ngành học")
plt.xlabel("Số lượng")

for bar in bars:
    plt.text(bar.get_width(), bar.get_y() + bar.get_height()/2,
             f"{int(bar.get_width())}", va="center")

plt.tight_layout()
plt.savefig("A2_nganh_hoc.png", dpi=300, bbox_inches="tight")
plt.show()


# A3 – Sinh viên (barh + số lượng)
plt.figure(figsize=(8, 6))
a3 = group_small(df.iloc[:, 2]).value_counts().sort_values()
bars = plt.barh(a3.index, a3.values)
plt.title("A3. Sinh viên")
plt.xlabel("Số lượng")

for bar in bars:
    plt.text(bar.get_width(), bar.get_y() + bar.get_height()/2,
             f"{int(bar.get_width())}", va="center")

plt.tight_layout()
plt.savefig("A3_sinh_vien.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("Data.xlsx")

b2_raw = df.iloc[:, 5].dropna()

time_slots = ["Sáng", "Trưa", "Chiều", "Tối"]
counts = {t: 0 for t in time_slots}

for ans in b2_raw:
    for t in time_slots:
        if t in str(ans):
            counts[t] += 1

b2 = pd.Series(counts)

plt.figure(figsize=(6, 5))
bars = plt.bar(b2.index, b2.values)
plt.title("B2. Thời điểm sử dụng Facebook trong ngày")
plt.xlabel("Thời điểm")
plt.ylabel("Số lượt xuất hiện")

for bar in bars:
    plt.text(bar.get_x() + bar.get_width()/2,
             bar.get_height(),
             int(bar.get_height()),
             ha="center", va="bottom")

plt.tight_layout()
plt.savefig("B2_thoi_diem_su_dung_phan_tich.png", dpi=300)
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_excel("Data.xlsx")

b4_raw = df.iloc[:, 7].dropna()

counts = {
    "Điện thoại": 0,
    "Laptop/PC": 0,
    "Máy tính bảng": 0
}

for ans in b4_raw:
    ans = str(ans).lower()

    if "điện thoại" in ans:
        counts["Điện thoại"] += 1
    if "laptop" in ans or "pc" in ans:
        counts["Laptop/PC"] += 1
    if "máy tính bảng" in ans or "tablet" in ans:
        counts["Máy tính bảng"] += 1

b4 = pd.Series(counts)
b4 = b4[b4 > 0]

plt.figure(figsize=(6, 6))
b4.plot(kind="pie", autopct="%1.1f%%", startangle=90)
plt.title("B4. Thiết bị sử dụng Facebook")
plt.ylabel("")
plt.tight_layout()
plt.savefig("B4_thiet_bi_su_dung_DUNG.png", dpi=300)
plt.show()
```

```python
import os
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# =========================
# 1) Load Excel (auto-pick)
# =========================
def pick_excel_file(default_name="Data.xlsx"):
    # ưu tiên file cùng thư mục chạy code
    if os.path.exists(default_name):
        return default_name

    # nếu không có, mở hộp thoại chọn file
    try:
        import tkinter as tk
        from tkinter import filedialog
        root = tk.Tk()
        root.withdraw()
        file_path = filedialog.askopenfilename(
            title="Chọn file Excel",
            filetypes=[("Excel files", "*.xlsx *.xls")]
        )
        if file_path:
            return file_path
    except Exception:
        pass

    raise FileNotFoundError(
        f"Không tìm thấy '{default_name}' trong thư mục hiện tại "
        f"({os.getcwd()}).\n"
        "👉 Hãy đặt file Data.xlsx cùng thư mục với code hoặc chọn file khi hộp thoại hiện lên."
    )

excel_path = pick_excel_file("Data.xlsx")
df = pd.read_excel(excel_path)

# =========================
# 2) Helper functions
# =========================
def find_col(prefix: str) -> str:
    for c in df.columns:
        if str(c).strip().startswith(prefix):
            return c
    raise KeyError(f"Không tìm thấy cột bắt đầu bằng '{prefix}'. Các cột hiện có: {list(df.columns)}")

def leading_num(x) -> int:
    m = re.match(r"\s*(\d+)\.", str(x))
    return int(m.group(1)) if m else 999

def value_counts_clean(s: pd.Series) -> pd.Series:
    s = s.astype(str).str.strip()
    s = s.replace({"nan": np.nan, "None": np.nan, "": np.nan})
    return s.value_counts(dropna=True)

def ensure_dir(d):
    os.makedirs(d, exist_ok=True)

def savefig(out_dir: str, fname: str):
    ensure_dir(out_dir)
    fp = os.path.join(out_dir, fname)
    plt.tight_layout()
    plt.savefig(fp, dpi=200, bbox_inches="tight")
    plt.close()
    return fp

def plot_count_and_pct(series: pd.Series,
                       title: str,
                       xlabel: str,
                       out_dir: str,
                       out_name: str,
                       sort_mode: str = "count_desc",
                       rotate: int = 20):
    """
    Vẽ biểu đồ kết hợp:
      - Cột: số lượng
      - Đường: tỷ lệ %
    sort_mode:
      - 'count_desc'  : theo số lượng giảm dần
      - 'leading_num' : theo thứ tự '1.', '2.'... nếu có
      - 'none'        : giữ nguyên value_counts (thực tế vẫn là count_desc)
    """
    counts = value_counts_clean(series)

    if sort_mode == "leading_num":
        counts = counts.sort_index(key=lambda idx: [leading_num(x) for x in idx])
    elif sort_mode == "count_desc":
        counts = counts.sort_values(ascending=False)

    total = counts.sum()
    pct = (counts / total * 100) if total > 0 else counts * 0

    x = np.arange(len(counts.index))

    fig, ax = plt.subplots(figsize=(10, 4))

    # Cột số lượng
    ax.bar(x, counts.values)
    ax.set_title(title)
    ax.set_xlabel(xlabel)
    ax.set_ylabel("Số lượng")

    ax.set_xticks(x)
    ax.set_xticklabels(counts.index, rotation=rotate, ha="right")

    # Đường tỷ lệ %
    ax2 = ax.twinx()
    ax2.plot(x, pct.values, marker="o")
    ax2.set_ylabel("Tỷ lệ (%)")
    ax2.set_ylim(0, max(100, float(pct.max()) * 1.2 if len(pct) else 100))

    # Ghi nhãn % trên điểm
    for xi, p in zip(x, pct.values):
        ax2.annotate(f"{p:.1f}%", (xi, p),
                     textcoords="offset points", xytext=(0, 6),
                     ha="center", fontsize=8)

    return savefig(out_dir, out_name)

# =========================
# 3) Columns mapping (A, B)
# =========================
col_gender = find_col("A1.")
col_major  = find_col("A2.")
col_year   = find_col("A3.")
col_gpa    = find_col("A4.")
col_b1     = find_col("B1.")
col_b2     = find_col("B2.")

# =========================
# 4) Generate charts
# =========================
out_dir = "plots_count_pct"
paths = []

paths.append(plot_count_and_pct(
    df[col_gender],
    "A1. Giới tính — Cột: số lượng | Đường: tỷ lệ (%)",
    "Giới tính",
    out_dir,
    "A1_gender_count_pct.png",
    sort_mode="count_desc",
    rotate=0
))

paths.append(plot_count_and_pct(
    df[col_major],
    "A2. Ngành học — Cột: số lượng | Đường: tỷ lệ (%)",
    "Ngành học",
    out_dir,
    "A2_major_count_pct.png",
    sort_mode="count_desc",
    rotate=25
))

paths.append(plot_count_and_pct(
    df[col_year],
    "A3. Năm học — Cột: số lượng | Đường: tỷ lệ (%)",
    "Năm học",
    out_dir,
    "A3_year_count_pct.png",
    sort_mode="leading_num",
    rotate=0
))

paths.append(plot_count_and_pct(
    df[col_gpa],
    "A4. Học lực — Cột: số lượng | Đường: tỷ lệ (%)",
    "Học lực",
    out_dir,
    "A4_gpa_count_pct.png",
    sort_mode="leading_num",
    rotate=20
))

paths.append(plot_count_and_pct(
    df[col_b1],
    "B1. Thời lượng dùng/ngày — Cột: số lượng | Đường: tỷ lệ (%)",
    "Nhóm thời lượng",
    out_dir,
    "B1_duration_count_pct.png",
    sort_mode="leading_num",
    rotate=20
))

paths.append(plot_count_and_pct(
    df[col_b2],
    "B2. Thời điểm sử dụng — Cột: số lượng | Đường: tỷ lệ (%)",
    "Thời điểm",
    out_dir,
    "B2_time_count_pct.png",
    sort_mode="count_desc",
    rotate=25
))

print("✅ Đã xuất các biểu đồ (Count + %):")
for p in paths:
    print(" -", os.path.abspath(p))

print(f"\n📁 Ảnh nằm trong thư mục: {os.path.abspath(out_dir)}")
```
